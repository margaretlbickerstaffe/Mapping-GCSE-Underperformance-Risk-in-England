```python
# import pandas
import pandas as pd
```

```python
# load in data from the Key Stage 4 Performance data (Department for Education)
secondarySchools = pd.read_csv("/content/drive/MyDrive/DISSERTATION/DataFiles/SecondarySchools/secondarySchoolsWithOutcomes.csv")
```

```python
# keep only England schools
secondarySchools = secondarySchools.dropna(subset=['Easting', 'Northing'])
# limit to secondary schools that are open and no duplicates
secondarySchools = secondarySchools[(secondarySchools['PhaseOfEducation (name)'] == 'Secondary') & (secondarySchools['EstablishmentStatus (name)'] == 'Open')].drop_duplicates(subset='URN').copy()
```

```python
# check columns
secondarySchools.columns.tolist()
```

```python
# check the types of schools --> looking for non selective (publicaly funded) schools
print(secondarySchools['AdmissionsPolicy (name)'].unique())
print(secondarySchools['TypeOfEstablishment (name)'].unique())
```

```python
# check if there are any PRU's
print(secondarySchools['TypeOfEstablishment (name)'].value_counts())
```

```python
# drop any grammar schools through Admissions Policy is Selective
secondarySchools = secondarySchools[secondarySchools['AdmissionsPolicy (name)'] != 'Selective']
```

```python
# check final length and save
len(secondarySchools)
```

```python
secondarySchools.to_csv("/content/drive/MyDrive/DISSERTATION/DataFiles/SecondarySchools/finalSecondarySchools.csv", index=False)
```
